#### Classes

- A Class is like a blueprint or template for creating  Objects.
- It defince attributes(data/properties) and methods (function/behaviors) that the object will have.

```python
class Car:
    def __init__(self,brand,color, speed):
        self.brand = brand  #Attribute
        self.color = color
        self.speed = speed
        
    def drive(self):  #Method
        print(f"{self.color} {self.brand} is driving.")
    def stop(self):
        print(f"{self.brand} car is stopped.")
    def car_speed(self):
        print(f"{self.brand} Maximum speed is {self.speed} km/h. ")
```

```python
my_car1 = Car("Audi A6", "Black", 250)
```

```python
my_car1.drive()
```

```python
my_car1.stop()
```

```python
my_car1.car_speed()
```

```python
my_car2 = Car("BMW M5 CS","Green",314) 
```

```python
my_car2.drive()
my_car2.car_speed()
my_car2.stop()
```

#### Inheritance
- Inheritance allows one class to use the properties ans methods of another class.
- The class that gives properties is the parent (base) class, and the one that inherits is the child (derived) class.
- promots code resuablitiy

```python
class Vehicle:   # Parent class
    def start(self):
        print("Vehicle started.")

class Car_t(Vehicle):  # Child class inherits Vehicle
    def honk(self):
        print("Car is honking.")

my_car = Car_t()
my_car.start()   # Inherited from Vehicle
my_car.honk()    # Defined in Car
```

```python
# Parent Class (Superclass)
class Vehicle_l:
    def __init__(self, brand):
        self.brand = brand
    def show_brand(self):
        print(f"This vehicle is a {self.brand}.")
    def drive(self):
        print("The vehicle is driving.")

# Child Class (Subclass) inheriting from Vehicle
class N_Car(Vehicle_l):
    def __init__(self,brand,speed):
    #call the parent class's constructor
        super().__init__(brand)
        self.speed = speed
    
    # Method overriding: A specialized 'drive' method for Car
    def drive(self):
        print(f"The {self.brand} car is driving on four wheels.")

# Another Child Class
class Motorcycle(Vehicle_l):
    def __init__(self, brand, has_sidecar):
        super().__init__(brand)
        self.has_sidecar = has_sidecar
    
    # This class uses the inherited 'drive' method from the parent
```

```python
my_v = Vehicle_l("Honda")
my_v.show_brand()
```

```python
# Create objects of the child classes
my_v2 = N_Car("Ford", 4)
my_motor = Motorcycle("Glammer", False)
my_v2.show_brand()
```

```python
my_motor.show_brand()
```

```python
my_v2.drive()
```

```python
my_motor.drive()
```

#### Encapsulation

- Encapsulation means hiding the internal details of an object and only exposing necessary parts.
- Achieved using access modifiers (public, protected, private).
- In Python:
     - public → accessible everywhere (default).
  - _protected → should not be accessed outside class (by convention).
  - __private → cannot be accessed directly outside the class.
  

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # Private attribute

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):  # Controlled access
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())   # ✅ 1500
# print(account.__balance)    # ❌ Error (private)
```
