# Class 12 Matrix Operations

```python
import numpy as np
```

## ------------------ MATRIX BASICS ------------------ #

### Define sample matrices

```python
# Define sample matrices
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

C = np.array([[2, 0, -1],
              [3, 4, 5]])

print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("Matrix C:\n", C)
```

## ------------------ MATRIX OPERATIONS ------------------ #

### 1. Addition & Subtraction

```python

# 1. Addition & Subtraction
print("\nA + B =\n", A + B)
print("A - B =\n", A - B)
```

### 2. Scalar Multiplication

```python
# 2. Scalar Multiplication
print("\n2A =\n", 2 * A)
```

### 3. Matrix Multiplication (A x B)

```python
# 3. Matrix Multiplication (A x B)
print("\nA x B =\n", np.dot(A, B))
```

```python
a = np.array([[1, 2, 3],[0, 1, 2]])
b = np.array([[1, 0],[2, 1], [3, 2]])
multiplication = np.dot(a, b)
print("Matrix of a : \n", a)
print("Matrix of b : \n", b)
print("Multiplicaton a * b : \n", multiplication)
```

```python
c = np.array([[1, 4, 0],[2, 5, 0],[3, 6, 0]])
d = np.array([[0, 0, 0],[0, 0, 0], [7 , 8, 9]])
multiplication = np.dot(c, d)
print("Matrix of c : \n", c)
print("Matrix of d : \n", d)
print("Multiplicaton c * d : \n", multiplication)
```

#### AB = AC

```python
e = np.array([[1, 4, 0],[2, 5, 0],[3, 6, 0]])
f = np.array([[3, 2, 1],[1, 2, 3], [4 , 5, 6]])
g = np.array([[3, 2, 1],[1, 2, 3], [7 , 8, 9]])
mul1 = np.dot(e, f)
mul2 = np.dot(e, g)
print("Matrix of e : \n", e)
print("Matrix of f : \n", f)
print("Matrix of g : \n", g)
print("Multiplicaton e * f : \n", mul1)
print("Multiplicaton e * g : \n", mul2)
```

### 4. Transpose of a Matrix

```python

# 4. Transpose of a Matrix
print("Matrix of A : \n",A)
print("Matrix of C : \n",C)
print("\nTranspose of A:\n", A.T)
print("Transpose of C:\n", C.T)
```

## ------------------ SPECIAL MATRICES ------------------ #

### 5. Identity Matrix

```python
# 5. Identity Matrix
I = np.eye(3, dtype=int)   # 2x2 identity
print("\nIdentity Matrix (I):\n", I)
```

### 6. Zero Matrix

```python
# 6. Zero Matrix
Z = np.zeros((3, 3), dtype=int)
print("\nZero Matrix:\n", Z)
```

### 7. Diagonal Matrix

```python
# 7. Diagonal Matrix
D = np.diag([1, 2, 3,4])
print("\nDiagonal Matrix:\n", D)
```

## ------------------ DETERMINANTS ------------------ #

### 8. Determinant of a Matrix

```python
# 8. Determinant of a Matrix
print("Matrix of A : \n",A)
det_A = np.linalg.det(A)
print("\nDeterminant of A:", det_A)
```

## ------------------ INVERSE ------------------ #

### 9. Inverse of a Matrix (only if determinant ≠ 0)

```python
# 9. Inverse of a Matrix (only if determinant ≠ 0)
print("Matrix of A : \n",A)
if np.linalg.det(A) != 0:
    inv_A = np.linalg.inv(A)
    print("\nInverse of A:\n", inv_A)
else:
    print("\nMatrix A is not invertible")
```

## ------------------ PROPERTIES ------------------ #

#### 10. Verify (A * A^(-1) = I)

```python
# 10. Verify (A * A^(-1) = I)
print("Matrix of A : \n",A)
print("\nInverse of A:\n", inv_A)
if np.linalg.det(A) != 0:
    print("\nA x A^(-1) =\n", np.dot(A, inv_A))
```

```python
print("Matrix of A : \n",A)
print("\nInverse of A:\n", inv_A)
#Verify 𝐴⋅𝐴^−1 = I
verify = np.dot(A, inv_A)
verify_rounded = np.round(verify) 
print("Check : \n",verify_rounded)
```

### 11. Symmetric and Skew-Symmetric parts


```python
# 11. Symmetric and Skew-Symmetric parts
print("Matrix of A : \n",A)

sym_A = (A + A.T) / 2
skew_A = (A - A.T) / 2
print("\nSymmetric part of A:\n", sym_A)
print("Skew-Symmetric part of A:\n", skew_A)
```

## ------------------ RANK & TRACE ------------------ #

### 12. Trace of a Matrix (sum of diagonal elements)

```python
# 12. Trace of a Matrix (sum of diagonal elements)
print("Matrix of A : \n",A)

print("\nTrace of A:", np.trace(A))
```

### 13. Rank of a Matrix

```python
# 13. Rank of a Matrix
print("Matrix of C : \n",C)
print("Rank of C:", np.linalg.matrix_rank(C))
```

## ------------------ EIGENVALUES ------------------ #

### 14. Eigenvalues & Eigenvectors

```python
# 14. Eigenvalues & Eigenvectors
print("Matrix of A : \n",A)
eigvals, eigvecs = np.linalg.eig(A)
print("\nEigenvalues of A:", eigvals)
print("Eigenvectors of A:\n", eigvecs)
```
