## Now Pratice 

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("==== Zomato sales analysis ==== ")
print("\n Sales data shape", sales_data.shape)
print("\n Sample data for 1st 3 restau: ", sales_data[:3])
print("\n Sales data (2021-2024): ", sales_data[:, 1:])
```

## Total sum

```python

# total sales per year
print("Total sales (2021-2024): ", np.sum(sales_data, axis=0))
year_total = np.sum(sales_data[:, 1:], axis=0)
print("Total sales per year: ", year_total)
# total sales per restaurant
print("Total sales per restaurant : ",np.sum(sales_data[:, 1:], axis=1))

print("Overall total sales : ", sum(year_total))
```

### Avg, min, max

```python
# Minimum sales per restaurant
min_sales = np.min(sales_data[:, 1:], axis=1)
print("Minimum sales per restaurant: ", min_sales)

# Maximum sales per year
max_sales = np.max(sales_data[:, 1:], axis=0)
print("Maximum sales per year: ", max_sales)

# Average sales per restaurant
avg_sales = np.mean(sales_data[:, 1:], axis=1)
print("Average of sales per restaurant: ", avg_sales)
```

```python
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(cumsum)

plt.figure(figsize=(8, 6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales across all restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

## Vector

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])

print("Vector addition: ", vector1 + vector2)

print("\n Multiplication vector", vector1 * vector2)

print("\nDot Product", np.dot(vector1, vector2))

angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)
```

```python
vector3 = np.array([[1,4],[2,5],[3,6]])
print("Vector 3: ", vector3)
vector4 = np.array([[2,3]])
print("Vector 4: ", vector4)
multi = vector3 * vector4
print("\n Multiplication vector", multi)
print("\n vector multiplication addition: ", np.sum(multi, axis=1))

#dot
dot_product = np.dot(vector3, vector4.T)
print("\n Dot Product: ", dot_product) 
```
