```python
#Drop Duplicate value remove

dup = pd.DataFrame({
    'A': [1, "jaga", "jaga", 4],
    'B': [5, np.nan,"jaga", 8],
    'C': [9, 10, "jaga", np.nan],
    'D': [3, np.nan, np.nan, np.nan],
    'E': [np.nan, np.nan, np.nan, np.nan]
})
```

```python
dup
```

```python
dup.drop_duplicates(subset=['A'])
```

```python
dup.drop_duplicates(subset=['A'],keep='first')
```

```python
dup.drop_duplicates(subset=['A'],keep='last')
```

```python
dup.drop_duplicates(subset=['A'],keep=False)
```
